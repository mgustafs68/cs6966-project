```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import textwrap

# ====== CONFIG ======
CLEAN_CSV = "/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/latent_labeling/latent_labels_clean.csv"
BUGGY_CSV = "/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/latent_labeling/latent_labels_buggy.csv"

OUT_DIR = Path("/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/viz")
OUT_DIR.mkdir(parents=True, exist_ok=True)
TOP_K = 5               # top latents per condition
WRAP_WIDTH = 42         # line wrap for snippet text
SNIPPET_CHARS = 140     # max characters from description to show


OUT_FIG = OUT_DIR / "fig_top5_latent_labels_clean_vs_buggy.png"
# ====== LOAD + CLEAN ======
def load_df(path: str, tag: str) -> pd.DataFrame:
    df = pd.read_csv(path).copy()
    df["pair"] = tag
    df["category"] = df.get("category", "unknown").fillna("unknown").astype(str).str.strip().str.lower()
    df["short_label"] = df.get("short_label", "").fillna("").astype(str).str.strip()
    df["description"] = df.get("description", "").fillna("").astype(str).str.strip()
    df["confidence"] = pd.to_numeric(df.get("confidence", np.nan), errors="coerce")
    df["is_parse_error"] = df["short_label"].str.lower().eq("parse_error")
    return df

clean = load_df(CLEAN_CSV, "clean_pair")
buggy = load_df(BUGGY_CSV, "buggy_pair")


# ====== SELECT TOP-K LATENTS ======
def top_k_latents(df: pd.DataFrame, k: int) -> pd.DataFrame:
    d = df.copy()
    # keep good rows
    d = d[~d["is_parse_error"]]
    d = d[~d["confidence"].isna()]
    d = d[d["short_label"] != ""]
    # optional: exclude unknown to get more meaningful labels
    # comment this out if you want unknowns included
    d = d[d["category"] != "unknown"]
    # pick top-k by confidence
    d = d.sort_values("confidence", ascending=False).head(k)
    return d

clean_top = top_k_latents(clean, TOP_K)
buggy_top = top_k_latents(buggy, TOP_K)

if len(clean_top) < TOP_K:
    print(f"Warning: clean_top has only {len(clean_top)} rows after filtering.")
if len(buggy_top) < TOP_K:
    print(f"Warning: buggy_top has only {len(buggy_top)} rows after filtering.")


# ====== BUILD DISPLAY TEXT ======
def make_snippet(desc: str, max_chars: int) -> str:
    desc = desc.strip().replace("\n", " ")
    if len(desc) > max_chars:
        desc = desc[:max_chars-1].rstrip() + "…"
    return desc

def row_to_display(row: pd.Series) -> str:
    lid = row.get("latent_idx", "")
    label = row.get("short_label", "")
    conf = row.get("confidence", np.nan)
    cat = row.get("category", "")
    snippet = make_snippet(row.get("description", ""), SNIPPET_CHARS)

    # Wrap snippet for readability
    wrapped_snip = "\n".join(textwrap.wrap(snippet, width=WRAP_WIDTH)) if snippet else "(no snippet)"
    # One card per latent
    return f"latent {lid}  |  {label}\ncat: {cat}  |  conf: {conf:.2f}\n{wrapped_snip}"

clean_cards = [row_to_display(r) for _, r in clean_top.iterrows()]
buggy_cards = [row_to_display(r) for _, r in buggy_top.iterrows()]


# ====== PLOT AS TABLE-LIKE FIGURE ======
# We'll draw text blocks in two columns (Clean vs Buggy), 5 rows each.
n_rows = max(len(clean_cards), len(buggy_cards), TOP_K)

plt.figure(figsize=(14, 10))  # adjust size for poster
ax = plt.gca()
ax.set_axis_off()

# Title
ax.text(0.5, 0.98, "Top-5 Latent Labels (Clean vs Buggy Crosscoder)",
        ha="center", va="top", fontsize=16, fontweight="bold")

# Column headers
ax.text(0.25, 0.93, "Clean pair", ha="center", va="top", fontsize=14, fontweight="bold")
ax.text(0.75, 0.93, "Buggy pair", ha="center", va="top", fontsize=14, fontweight="bold")

# Layout parameters
top_y = 0.90
bottom_y = 0.06
row_h = (top_y - bottom_y) / n_rows

# Draw rows
for i in range(n_rows):
    y = top_y - i * row_h

    # row separator line
    ax.plot([0.05, 0.95], [y - row_h + 0.01, y - row_h + 0.01], linewidth=0.8)

    # Clean card
    if i < len(clean_cards):
        ax.text(0.05, y, clean_cards[i], ha="left", va="top", fontsize=10)

    # Buggy card
    if i < len(buggy_cards):
        ax.text(0.55, y, buggy_cards[i], ha="left", va="top", fontsize=10)

# Vertical separator between columns
ax.plot([0.5, 0.5], [bottom_y, top_y], linewidth=1.0)

plt.tight_layout()
plt.savefig(OUT_FIG, dpi=220)
plt.show()

print("Saved figure:", OUT_FIG)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import textwrap

# ====== CONFIG ======
CLEAN_CSV = "/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/latent_labeling/latent_labels_clean.csv"
BUGGY_CSV = "/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/latent_labeling/latent_labels_buggy.csv"

OUT_DIR = Path("/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/viz")
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_FIG = OUT_DIR / "fig_top5_latents_heatmap_table.png"

TOP_K = 5
WRAP_WIDTH = 34         # narrower to fit in a proper table cell
SNIPPET_CHARS = 110     # shorter snippet for scanning

# ====== LOAD + CLEAN ======
def load_df(path: str, tag: str) -> pd.DataFrame:
    df = pd.read_csv(path).copy()
    df["pair"] = tag
    df["category"] = df.get("category", "unknown").fillna("unknown").astype(str).str.strip().str.lower()
    df["short_label"] = df.get("short_label", "").fillna("").astype(str).str.strip()
    df["description"] = df.get("description", "").fillna("").astype(str).str.strip()
    df["confidence"] = pd.to_numeric(df.get("confidence", np.nan), errors="coerce")
    df["is_parse_error"] = df["short_label"].str.lower().eq("parse_error")
    return df

clean = load_df(CLEAN_CSV, "clean_pair")
buggy = load_df(BUGGY_CSV, "buggy_pair")

# ====== SELECT TOP-K LATENTS ======
def top_k_latents(df: pd.DataFrame, k: int) -> pd.DataFrame:
    d = df.copy()
    d = d[~d["is_parse_error"]]
    d = d[~d["confidence"].isna()]
    d = d[d["short_label"] != ""]
    d = d[d["category"] != "unknown"]          # comment out if you need to include unknowns
    d = d.sort_values("confidence", ascending=False).head(k)
    return d.reset_index(drop=True)

clean_top = top_k_latents(clean, TOP_K)
buggy_top = top_k_latents(buggy, TOP_K)

if len(clean_top) < TOP_K:
    print(f"Warning: clean_top has only {len(clean_top)} rows after filtering.")
if len(buggy_top) < TOP_K:
    print(f"Warning: buggy_top has only {len(buggy_top)} rows after filtering.")

# ====== TEXT HELPERS ======
def make_snippet(desc: str, max_chars: int) -> str:
    desc = desc.strip().replace("\n", " ")
    if len(desc) > max_chars:
        desc = desc[:max_chars-1].rstrip() + "…"
    return desc

def wrap(s: str, w: int) -> str:
    return "\n".join(textwrap.wrap(s, width=w)) if s else ""

def build_cell_text(row: pd.Series) -> str:
    lid = row.get("latent_idx", "")
    label = row.get("short_label", "")
    conf = row.get("confidence", np.nan)
    cat = row.get("category", "")
    snippet = wrap(make_snippet(row.get("description", ""), SNIPPET_CHARS), WRAP_WIDTH)
    return f"{label}\n(latent {lid}, {cat})\n{snippet}"

# ====== BUILD TABLE MATRIX ======
# rows: rank 1..K
# cols: Clean | Buggy (each cell shows label+snippet)
row_labels = [f"Rank {i+1}" for i in range(TOP_K)]
col_labels = ["Clean pair", "Buggy pair"]

cell_text = []
cell_conf = []
for i in range(TOP_K):
    c_row = clean_top.iloc[i] if i < len(clean_top) else pd.Series({})
    b_row = buggy_top.iloc[i] if i < len(buggy_top) else pd.Series({})
    cell_text.append([
        build_cell_text(c_row) if len(c_row) else "(missing)",
        build_cell_text(b_row) if len(b_row) else "(missing)"
    ])
    cell_conf.append([
        float(c_row.get("confidence", np.nan)) if len(c_row) else np.nan,
        float(b_row.get("confidence", np.nan)) if len(b_row) else np.nan
    ])

cell_conf = np.array(cell_conf, dtype=float)  # shape [K,2]

# ====== PLOT: heatmap background + real table overlay ======
fig, ax = plt.subplots(figsize=(12, 7.5))

# Heatmap encodes confidence (NaNs shown as 0 but we will hatch/label them)
heat = np.nan_to_num(cell_conf, nan=0.0)

im = ax.imshow(heat, aspect="auto")  # no explicit colors

# Ticks/labels
ax.set_xticks(np.arange(len(col_labels)))
ax.set_xticklabels(col_labels, fontsize=12, fontweight="bold")
ax.set_yticks(np.arange(len(row_labels)))
ax.set_yticklabels(row_labels, fontsize=11, fontweight="bold")

# Gridlines for table look
ax.set_xticks(np.arange(-.5, len(col_labels), 1), minor=True)
ax.set_yticks(np.arange(-.5, TOP_K, 1), minor=True)
ax.grid(which="minor", linestyle='-', linewidth=1.2)
ax.tick_params(which="minor", bottom=False, left=False)

# Put the text inside each cell
for i in range(TOP_K):
    for j in range(2):
        txt = cell_text[i][j]
        conf = cell_conf[i, j]
        # Add confidence on first line corner for quick scanning
        conf_str = "conf: NA" if np.isnan(conf) else f"conf: {conf:.2f}"
        full_txt = conf_str + "\n" + txt
        ax.text(j, i, full_txt, ha="center", va="center", fontsize=9)

# Colorbar
cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Confidence", rotation=90)

ax.set_title("Top-5 Latent Labels (Clean vs Buggy) — Confidence Heatmap + Snippets",
             fontsize=14, fontweight="bold", pad=14)

plt.tight_layout()
plt.savefig(OUT_FIG, dpi=250)
plt.show()

print("Saved figure:", OUT_FIG)
```

```python
"""
poster_figures.py
=================
Generates poster-ready figures from crosscoder latent label CSVs.

Run:
    python poster_figures.py

Outputs (saved to ./figures/):
    fig1_category_comparison.pdf   — grouped bar chart: category counts clean vs buggy
    fig2_latent_heatmap.pdf        — heatmap of all 50 latents x {category, confidence}
    fig3_valid_latents_table.pdf   — annotated table of non-unknown latents with confidence
"""

import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.colors import LinearSegmentedColormap
from pathlib import Path

# ── Output dir ────────────────────────────────────────────────────────────────
OUT = Path("outputs/viz2")
OUT.mkdir(exist_ok=True)

# ── Poster typography ─────────────────────────────────────────────────────────
matplotlib.rcParams.update({
    "font.family":      "sans-serif",
    "font.size":        11,
    "axes.titlesize":   13,
    "axes.labelsize":   11,
    "xtick.labelsize":  10,
    "ytick.labelsize":  10,
    "axes.spines.top":  False,
    "axes.spines.right":False,
    "figure.dpi":       150,
    "savefig.dpi":      300,
    "savefig.bbox":     "tight",
    "savefig.pad_inches": 0.05,
})

# ── Colors ────────────────────────────────────────────────────────────────────
CLEAN_COLOR  = "#4878CF"   # blue
BUGGY_COLOR  = "#E8604C"   # red-orange
UNKNOWN_COLOR = "#CCCCCC"  # light gray for unknown/parse-error

CATEGORY_COLORS = {
    "semantic":    "#2196F3",
    "proper_noun": "#9C27B0",
    "formatting":  "#FF9800",
    "punctuation": "#009688",
    "stopword":    "#795548",
    "unknown":     "#BDBDBD",
}

# ── Load data ─────────────────────────────────────────────────────────────────
clean = pd.read_csv("/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/latent_labeling/latent_labels_clean.csv")
buggy = pd.read_csv("/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/latent_labeling/latent_labels_buggy.csv")

# Treat parse_error rows as unknown with confidence=0
clean["category"]   = clean["category"].fillna("unknown")
buggy["category"]   = buggy["category"].fillna("unknown")
clean["confidence"] = clean["confidence"].fillna(0.0)
buggy["confidence"] = buggy["confidence"].fillna(0.0)

# Separate valid (successfully labeled) from unknown/parse-error
clean_valid = clean[clean["short_label"] != "parse_error"].copy()
buggy_valid = buggy[buggy["short_label"] != "parse_error"].copy()

print(f"Clean: {len(clean)} total, {len(clean_valid)} valid")
print(f"Buggy: {len(buggy)} total, {len(buggy_valid)} valid")


# ══════════════════════════════════════════════════════════════════════════════
# Figure 1 — Category comparison bar chart
# ══════════════════════════════════════════════════════════════════════════════

def fig1_category_comparison():
    all_cats = ["semantic", "proper_noun", "formatting", "punctuation", "stopword", "unknown"]

    clean_counts = clean["category"].value_counts().reindex(all_cats, fill_value=0)
    buggy_counts = buggy["category"].value_counts().reindex(all_cats, fill_value=0)

    x      = np.arange(len(all_cats))
    width  = 0.35

    fig, ax = plt.subplots(figsize=(8, 4))

    bars_clean = ax.bar(x - width/2, clean_counts.values, width,
                        color=CLEAN_COLOR, alpha=0.85, label="Clean crosscoder")
    bars_buggy = ax.bar(x + width/2, buggy_counts.values, width,
                        color=BUGGY_COLOR, alpha=0.85, label="Buggy crosscoder")

    # Value labels on bars
    for bar in list(bars_clean) + list(bars_buggy):
        h = bar.get_height()
        if h > 0:
            ax.text(bar.get_x() + bar.get_width() / 2, h + 0.3,
                    str(int(h)), ha="center", va="bottom", fontsize=9)

    ax.set_xticks(x)
    ax.set_xticklabels([c.replace("_", "\n") for c in all_cats])
    ax.set_ylabel("Number of latents")
    ax.set_title("Latent category distribution — Clean vs Buggy crosscoder\n(top 50 latents by sycophancy contrast)")
    ax.legend(frameon=False)
    ax.set_ylim(0, max(clean_counts.max(), buggy_counts.max()) + 5)

    plt.tight_layout()
    fig.savefig(OUT / "fig1_category_comparison.pdf")
    fig.savefig(OUT / "fig1_category_comparison.png")
    plt.close(fig)
    print("Saved fig1")


# ══════════════════════════════════════════════════════════════════════════════
# Figure 2 — Heatmap: all 50 latents, rows = latent index, cols = crosscoder
# Each cell colored by category, intensity = confidence
# ══════════════════════════════════════════════════════════════════════════════

def fig2_latent_heatmap():
    """
    Side-by-side heatmap: left = clean crosscoder, right = buggy crosscoder.
    Each row = one of the 50 latents (by position in the ranked list, not latent_idx).
    Cell color = category color, alpha = confidence.
    Unknown/parse-error cells are gray.
    """
    n = 50

    def make_rgba_matrix(df):
        """Return (n, 4) RGBA array for heatmap cells."""
        rgba = np.zeros((n, 4))
        for i, row in df.reset_index(drop=True).iterrows():
            cat  = row["category"] if row["short_label"] != "parse_error" else "unknown"
            conf = float(row["confidence"]) if row["short_label"] != "parse_error" else 0.0
            hex_color = CATEGORY_COLORS.get(cat, CATEGORY_COLORS["unknown"])
            r, g, b = matplotlib.colors.to_rgb(hex_color)
            # Alpha: unknown → 0.15, others scaled by confidence (min 0.3 so always visible)
            alpha = max(0.15, conf) if cat != "unknown" else 0.15
            rgba[i] = [r, g, b, alpha]
        return rgba

    clean_rgba = make_rgba_matrix(clean)
    buggy_rgba = make_rgba_matrix(buggy)

    # Labels for y-axis: position + latent_idx + short_label
    def row_labels(df):
        labels = []
        for _, row in df.reset_index(drop=True).iterrows():
            sl = row["short_label"] if row["short_label"] != "parse_error" else "—"
            labels.append(f"{row['latent_idx']:>4}  {sl}")
        return labels

    clean_labels = row_labels(clean)
    buggy_labels = row_labels(buggy)

    fig, axes = plt.subplots(1, 2, figsize=(11, 14),
                             gridspec_kw={"wspace": 0.55})

    for ax, rgba, labels, title in [
        (axes[0], clean_rgba, clean_labels, "Clean crosscoder"),
        (axes[1], buggy_rgba, buggy_labels, "Buggy crosscoder"),
    ]:
        # Draw cells manually as colored rectangles
        for i in range(n):
            rect = mpatches.FancyBboxPatch(
                (0, n - 1 - i), 1, 0.88,
                boxstyle="round,pad=0.04",
                facecolor=rgba[i, :3],
                alpha=float(rgba[i, 3]),
                edgecolor="white", linewidth=0.5,
            )
            ax.add_patch(rect)

            # Confidence text inside cell (skip if unknown)
            cat = clean["category"].iloc[i] if ax == axes[0] else buggy["category"].iloc[i]
            conf = clean["confidence"].iloc[i] if ax == axes[0] else buggy["confidence"].iloc[i]
            sl   = clean["short_label"].iloc[i] if ax == axes[0] else buggy["short_label"].iloc[i]
            if sl != "parse_error" and cat != "unknown" and conf > 0:
                ax.text(0.5, n - 1 - i + 0.44, f"{conf:.0%}",
                        ha="center", va="center", fontsize=7,
                        color="white" if conf > 0.5 else "#333333",
                        fontweight="bold")

        ax.set_xlim(0, 1)
        ax.set_ylim(0, n)
        ax.set_yticks(np.arange(n) + 0.44)
        ax.set_yticklabels(labels[::-1], fontsize=7, fontfamily="monospace")
        ax.set_xticks([])
        ax.set_title(title, fontsize=12, pad=8)
        ax.spines["left"].set_visible(False)
        ax.spines["bottom"].set_visible(False)

    # Legend
    legend_patches = [
        mpatches.Patch(color=CATEGORY_COLORS[cat], label=cat.replace("_", " "))
        for cat in CATEGORY_COLORS
    ]
    fig.legend(handles=legend_patches, loc="lower center", ncol=3,
               frameon=False, fontsize=9, title="Category",
               bbox_to_anchor=(0.5, 0.0))

    fig.suptitle("Top-50 sycophancy-contrast latents — label and confidence",
                 fontsize=13, y=1.005)

    plt.tight_layout(rect=[0, 0.06, 1, 1])
    fig.savefig(OUT / "fig2_latent_heatmap.pdf")
    fig.savefig(OUT / "fig2_latent_heatmap.png")
    plt.close(fig)
    print("Saved fig2")


# ══════════════════════════════════════════════════════════════════════════════
# Figure 3 — Clean table of valid (non-unknown) latents side by side
# ══════════════════════════════════════════════════════════════════════════════

def fig3_valid_latents_table():
    """
    Two-column table side by side: clean (left), buggy (right).
    Rows = valid labeled latents only (excluding unknown and parse_error).
    Columns: latent_idx | short_label | category | confidence
    Cell background = category color at confidence alpha.
    """
    def prep(df):
        valid = df[
            (df["short_label"] != "parse_error") &
            (df["category"] != "unknown")
        ][["latent_idx", "short_label", "category", "confidence"]].copy()
        valid = valid.reset_index(drop=True)
        valid["conf_pct"] = (valid["confidence"] * 100).round(0).astype(int).astype(str) + "%"
        return valid

    c = prep(clean)
    b = prep(buggy)

    col_labels = ["Latent", "Label", "Category", "Conf"]
    col_widths  = [0.12,      0.30,    0.30,       0.12]

    def draw_table(ax, df, title):
        n_rows = len(df)
        n_cols = len(col_labels)
        total_rows = n_rows + 1  # +1 for header

        ax.set_xlim(0, sum(col_widths))
        ax.set_ylim(0, total_rows)
        ax.axis("off")
        ax.set_title(title, fontsize=11, pad=6, fontweight="bold")

        # Header
        x = 0
        for j, (label, w) in enumerate(zip(col_labels, col_widths)):
            ax.add_patch(mpatches.Rectangle(
                (x, n_rows), w, 1,
                facecolor="#2C3E50", edgecolor="white", linewidth=0.5
            ))
            ax.text(x + w/2, n_rows + 0.5, label,
                    ha="center", va="center", fontsize=10,
                    color="white", fontweight="bold")
            x += w

        # Data rows
        for i, row in df.iterrows():
            cat   = row["category"]
            conf  = row["confidence"]
            color = CATEGORY_COLORS.get(cat, CATEGORY_COLORS["unknown"])
            alpha = max(0.2, conf * 0.6)  # gentle background, text stays readable
            row_y = n_rows - 1 - i

            values = [
                str(int(row["latent_idx"])),
                row["short_label"],
                row["category"].replace("_", " "),
                row["conf_pct"],
            ]

            x = 0
            for j, (val, w) in enumerate(zip(values, col_widths)):
                bg = color if j == 2 else ("#F5F5F5" if i % 2 == 0 else "white")
                bg_alpha = alpha if j == 2 else 1.0
                ax.add_patch(mpatches.Rectangle(
                    (x, row_y), w, 1,
                    facecolor=bg, alpha=bg_alpha,
                    edgecolor="#DDDDDD", linewidth=0.3
                ))
                text_color = "white" if (j == 2 and conf > 0.7) else "#222222"
                ax.text(x + w/2, row_y + 0.5, val,
                        ha="center", va="center", fontsize=10,
                        color=text_color)
                x += w

    fig, axes = plt.subplots(1, 2, figsize=(11, max(len(c), len(b)) * 0.35 + 1.5))

    draw_table(axes[0], c, f"Clean crosscoder  ({len(c)} valid latents)")
    draw_table(axes[1], b, f"Buggy crosscoder  ({len(b)} valid latents)")

    fig.suptitle("Valid labeled latents — category and confidence",
                 fontsize=13, y=1.02)

    # Legend
    legend_patches = [
        mpatches.Patch(color=CATEGORY_COLORS[cat], label=cat.replace("_", " "))
        for cat in CATEGORY_COLORS if cat != "unknown"
    ]
    fig.legend(handles=legend_patches, loc="lower center", ncol=5,
               frameon=False, fontsize=8.5, bbox_to_anchor=(0.5, -0.04))

    plt.tight_layout()
    fig.savefig(OUT / "fig3_valid_latents_table.pdf")
    fig.savefig(OUT / "fig3_valid_latents_table.png")
    plt.close(fig)
    print("Saved fig3")


# ══════════════════════════════════════════════════════════════════════════════
# Figure 4 — Confidence distribution by category (violin / strip)
# ══════════════════════════════════════════════════════════════════════════════

def fig4_confidence_by_category():
    """
    Horizontal strip plot showing confidence distribution per category,
    clean and buggy overlaid. Only valid (non-parse-error) rows included.
    """
    cv = clean_valid[clean_valid["category"] != "unknown"].copy()
    bv = buggy_valid[buggy_valid["category"] != "unknown"].copy()

    all_cats = sorted(set(cv["category"].tolist() + bv["category"].tolist()))

    fig, ax = plt.subplots(figsize=(7, len(all_cats) * 0.7 + 1.2))

    for i, cat in enumerate(all_cats):
        y = len(all_cats) - 1 - i
        cc = cv[cv["category"] == cat]["confidence"].values
        bc = bv[bv["category"] == cat]["confidence"].values

        # Jitter
        rng = np.random.default_rng(42)
        if len(cc):
            jitter = rng.uniform(-0.12, 0.12, len(cc))
            ax.scatter(cc, [y + 0.15] * len(cc) + jitter,
                       color=CLEAN_COLOR, s=55, alpha=0.85, zorder=3,
                       label="Clean" if i == 0 else "_")
        if len(bc):
            jitter = rng.uniform(-0.12, 0.12, len(bc))
            ax.scatter(bc, [y - 0.15] * len(bc) + jitter,
                       color=BUGGY_COLOR, s=55, alpha=0.85, zorder=3,
                       marker="D",
                       label="Buggy" if i == 0 else "_")

        # Mean lines
        if len(cc):
            ax.plot([cc.mean(), cc.mean()], [y + 0.02, y + 0.28],
                    color=CLEAN_COLOR, linewidth=2, zorder=4)
        if len(bc):
            ax.plot([bc.mean(), bc.mean()], [y - 0.28, y - 0.02],
                    color=BUGGY_COLOR, linewidth=2, zorder=4)

        # Horizontal separator
        ax.axhline(y + 0.5, color="#EEEEEE", linewidth=0.8, zorder=1)

    ax.set_yticks(range(len(all_cats)))
    ax.set_yticklabels([c.replace("_", " ") for c in reversed(all_cats)])
    ax.set_xlim(0, 1.05)
    ax.set_xlabel("Confidence")
    ax.set_title("Confidence by category — valid latents only\n(circles = clean, diamonds = buggy; vertical bar = mean)")
    ax.legend(frameon=False, loc="lower right")
    ax.axvline(0.5, color="#CCCCCC", linewidth=0.8, linestyle="--")
    ax.text(0.5, -0.6, "0.5", ha="center", va="top", fontsize=8, color="#999999")

    plt.tight_layout()
    fig.savefig(OUT / "fig4_confidence_by_category.pdf")
    fig.savefig(OUT / "fig4_confidence_by_category.png")
    plt.close(fig)
    print("Saved fig4")


# ══════════════════════════════════════════════════════════════════════════════
# Run all
# ══════════════════════════════════════════════════════════════════════════════

if __name__ == "__main__":
    fig1_category_comparison()
    fig2_latent_heatmap()
    fig3_valid_latents_table()
    fig4_confidence_by_category()
    print(f"\nAll figures saved to ./{OUT}/")
    print("Use the .pdf versions for the poster (vector, crisp at any size).")
    print("Use the .png versions for quick preview.")
```

```python
"""
poster_figures.py
=================
Generates poster-ready figures from crosscoder latent label CSVs.

Run:
    python poster_figures.py

Outputs saved to OUT_DIR:
    fig1_category_comparison.png/pdf
    fig2_latent_heatmap.png/pdf
    fig3_valid_latents_table.png/pdf
    fig4_confidence_by_category.png/pdf
"""

import textwrap
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from pathlib import Path

# ====== CONFIG ======
CLEAN_CSV = "/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/latent_labeling/latent_labels_clean.csv"
BUGGY_CSV = "/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/latent_labeling/latent_labels_buggy.csv"

OUT_DIR = Path("/uufs/chpc.utah.edu/common/home/u1528744/interpretability/cs6966-project/outputs/viz")
OUT_DIR.mkdir(parents=True, exist_ok=True)

TOP_K         = 5
WRAP_WIDTH    = 34
SNIPPET_CHARS = 110

# ── Poster typography ─────────────────────────────────────────────────────────
matplotlib.rcParams.update({
    "font.family":       "sans-serif",
    "font.size":         11,
    "axes.titlesize":    13,
    "axes.labelsize":    11,
    "xtick.labelsize":   10,
    "ytick.labelsize":   10,
    "axes.spines.top":   False,
    "axes.spines.right": False,
    "figure.dpi":        150,
    "savefig.dpi":       300,
    "savefig.bbox":      "tight",
    "savefig.pad_inches": 0.05,
})

# ── Colors ────────────────────────────────────────────────────────────────────
CLEAN_COLOR   = "#4878CF"
BUGGY_COLOR   = "#E8604C"

CATEGORY_COLORS = {
    "semantic":    "#2196F3",
    "proper_noun": "#9C27B0",
    "formatting":  "#FF9800",
    "punctuation": "#009688",
    "stopword":    "#795548",
    "unknown":     "#BDBDBD",
}

# ── Load & normalise ──────────────────────────────────────────────────────────
clean = pd.read_csv(CLEAN_CSV)
buggy = pd.read_csv(BUGGY_CSV)

for df in (clean, buggy):
    df["category"]   = df["category"].fillna("unknown")
    df["confidence"] = df["confidence"].fillna(0.0)

clean_valid = clean[clean["short_label"] != "parse_error"].copy()
buggy_valid = buggy[buggy["short_label"] != "parse_error"].copy()

print(f"Clean: {len(clean)} total, {len(clean_valid)} valid")
print(f"Buggy: {len(buggy)} total, {len(buggy_valid)} valid")


# ══════════════════════════════════════════════════════════════════════════════
# Figure 1 — Category comparison grouped bar chart
# ══════════════════════════════════════════════════════════════════════════════

def fig1_category_comparison():
    all_cats = ["semantic", "proper_noun", "formatting", "punctuation", "stopword", "unknown"]

    clean_counts = clean["category"].value_counts().reindex(all_cats, fill_value=0)
    buggy_counts = buggy["category"].value_counts().reindex(all_cats, fill_value=0)

    x     = np.arange(len(all_cats))
    width = 0.35

    fig, ax = plt.subplots(figsize=(8, 4))

    bars_c = ax.bar(x - width/2, clean_counts.values, width,
                    color=CLEAN_COLOR, alpha=0.85, label="Clean crosscoder")
    bars_b = ax.bar(x + width/2, buggy_counts.values, width,
                    color=BUGGY_COLOR, alpha=0.85, label="Buggy crosscoder")

    for bar in list(bars_c) + list(bars_b):
        h = bar.get_height()
        if h > 0:
            ax.text(bar.get_x() + bar.get_width() / 2, h + 0.3,
                    str(int(h)), ha="center", va="bottom", fontsize=9)

    ax.set_xticks(x)
    ax.set_xticklabels([c.replace("_", "\n") for c in all_cats])
    ax.set_ylabel("Number of latents")
    ax.set_title("Latent category distribution — Clean vs Buggy crosscoder\n"
                 "(top 50 latents by sycophancy contrast)")
    ax.legend(frameon=False)
    ax.set_ylim(0, max(clean_counts.max(), buggy_counts.max()) + 5)

    plt.tight_layout()
    for ext in ("pdf", "png"):
        fig.savefig(OUT_DIR / f"fig1_category_comparison.{ext}")
    plt.close(fig)
    print("Saved fig1")


# ══════════════════════════════════════════════════════════════════════════════
# Figure 2 — Full heatmap: 50 latents × 2 crosscoders
# Color = category, opacity = confidence
# ══════════════════════════════════════════════════════════════════════════════

def fig2_latent_heatmap():
    n = len(clean)  # typically 50

    def make_rgba(df):
        rgba = np.zeros((n, 4))
        for i, row in df.reset_index(drop=True).iterrows():
            is_parse = row["short_label"] == "parse_error"
            cat  = "unknown" if is_parse else row["category"]
            conf = 0.0       if is_parse else float(row["confidence"])
            r, g, b = matplotlib.colors.to_rgb(
                CATEGORY_COLORS.get(cat, CATEGORY_COLORS["unknown"])
            )
            alpha = 0.15 if cat == "unknown" else max(0.3, conf)
            rgba[i] = [r, g, b, alpha]
        return rgba

    def row_labels(df):
        out = []
        for _, row in df.reset_index(drop=True).iterrows():
            sl = "—" if row["short_label"] == "parse_error" else row["short_label"]
            out.append(f"{row['latent_idx']:>4}  {sl}")
        return out

    clean_rgba = make_rgba(clean)
    buggy_rgba = make_rgba(buggy)

    fig, axes = plt.subplots(1, 2, figsize=(11, 14),
                             gridspec_kw={"wspace": 0.55})

    panels = [
        (axes[0], clean_rgba, row_labels(clean), clean, "Clean crosscoder"),
        (axes[1], buggy_rgba, row_labels(buggy), buggy, "Buggy crosscoder"),
    ]

    for ax, rgba, labels, df_src, title in panels:
        for i in range(n):
            ax.add_patch(mpatches.FancyBboxPatch(
                (0, n - 1 - i), 1, 0.88,
                boxstyle="round,pad=0.04",
                facecolor=rgba[i, :3],
                alpha=float(rgba[i, 3]),
                edgecolor="white", linewidth=0.5,
            ))
            row = df_src.iloc[i]
            if row["short_label"] != "parse_error" and row["category"] != "unknown":
                conf = float(row["confidence"])
                ax.text(0.5, n - 1 - i + 0.44, f"{conf:.0%}",
                        ha="center", va="center", fontsize=7,
                        color="white" if conf > 0.5 else "#333333",
                        fontweight="bold")

        ax.set_xlim(0, 1)
        ax.set_ylim(0, n)
        ax.set_yticks(np.arange(n) + 0.44)
        ax.set_yticklabels(labels[::-1], fontsize=7, fontfamily="monospace")
        ax.set_xticks([])
        ax.set_title(title, fontsize=12, pad=8)
        ax.spines["left"].set_visible(False)
        ax.spines["bottom"].set_visible(False)

    legend_patches = [
        mpatches.Patch(color=CATEGORY_COLORS[cat], label=cat.replace("_", " "))
        for cat in CATEGORY_COLORS
    ]
    fig.legend(handles=legend_patches, loc="lower center", ncol=3,
               frameon=False, fontsize=9, title="Category",
               bbox_to_anchor=(0.5, 0.0))
    fig.suptitle("Top-50 sycophancy-contrast latents — label and confidence",
                 fontsize=13, y=1.005)

    plt.tight_layout(rect=[0, 0.06, 1, 1])
    for ext in ("pdf", "png"):
        fig.savefig(OUT_DIR / f"fig2_latent_heatmap.{ext}")
    plt.close(fig)
    print("Saved fig2")


# ══════════════════════════════════════════════════════════════════════════════
# Figure 3 — Annotated table of valid (non-unknown) latents
# ══════════════════════════════════════════════════════════════════════════════

def fig3_valid_latents_table():
    def prep(df):
        valid = df[
            (df["short_label"] != "parse_error") &
            (df["category"] != "unknown")
        ][["short_label", "category", "confidence"]].copy()
        valid = (valid
                 .sort_values("confidence", ascending=False)
                 .reset_index(drop=True))
        valid["conf_pct"] = (
            (valid["confidence"] * 100).round(0).astype(int).astype(str) + "%"
        )
        return valid

    c = prep(clean)
    b = prep(buggy)

    ROW_H      = 0.62
    col_labels = ["Label", "Category", "Conf"]
    col_widths  = [0.42,    0.38,       0.16]

    def draw_table(ax, df, title):
        n_rows  = len(df)
        total_h = n_rows * ROW_H + ROW_H
        ax.set_xlim(0, sum(col_widths))
        ax.set_ylim(0, total_h)
        ax.axis("off")
        ax.set_title(title, fontsize=12, pad=4, fontweight="bold")

        # Header row
        header_y = n_rows * ROW_H
        x = 0
        for label, w in zip(col_labels, col_widths):
            ax.add_patch(mpatches.Rectangle(
                (x, header_y), w, ROW_H,
                facecolor="#2C3E50", edgecolor="white", linewidth=0.4
            ))
            ax.text(x + w/2, header_y + ROW_H * 0.5, label,
                    ha="center", va="center", fontsize=12,
                    color="white", fontweight="bold")
            x += w

        # Data rows
        for i, (_, row) in enumerate(df.iterrows()):
            cat   = row["category"]
            conf  = row["confidence"]
            color = CATEGORY_COLORS.get(cat, CATEGORY_COLORS["unknown"])
            alpha = max(0.2, conf * 0.6)
            row_y = (n_rows - 1 - i) * ROW_H
            values = [
                row["short_label"],
                cat.replace("_", " "),
                row["conf_pct"],
            ]
            x = 0
            for j, (val, w) in enumerate(zip(values, col_widths)):
                bg       = color if j == 1 else ("#F5F5F5" if i % 2 == 0 else "white")
                bg_alpha = alpha if j == 1 else 1.0
                ax.add_patch(mpatches.Rectangle(
                    (x, row_y), w, ROW_H,
                    facecolor=bg, alpha=bg_alpha,
                    edgecolor="#DDDDDD", linewidth=0.25
                ))
                text_color = "white" if (j == 1 and conf > 0.7) else "#222222"
                ax.text(x + w/2, row_y + ROW_H * 0.5, val,
                        ha="center", va="center", fontsize=12,
                        color=text_color)
                x += w

    fig, axes = plt.subplots(
        1, 2,
        figsize=(11, max(len(c), len(b)) * ROW_H * 0.42 + 1.2),
        gridspec_kw={"wspace": 0.08}
    )
    draw_table(axes[0], c, f"Clean crosscoder  ({len(c)} valid latents)")
    draw_table(axes[1], b, f"Buggy crosscoder  ({len(b)} valid latents)")

    fig.suptitle("Valid labeled latents — category and confidence",
                 fontsize=13, y=1.01)

    legend_patches = [
        mpatches.Patch(color=CATEGORY_COLORS[cat], label=cat.replace("_", " "))
        for cat in CATEGORY_COLORS if cat != "unknown"
    ]
    fig.legend(handles=legend_patches, loc="lower center", ncol=5,
               frameon=False, fontsize=11, bbox_to_anchor=(0.5, -0.02))

    plt.tight_layout(pad=0.4)


# ══════════════════════════════════════════════════════════════════════════════
# Figure 4 — Confidence by category strip plot
# ══════════════════════════════════════════════════════════════════════════════

def fig4_confidence_by_category():
    cv = clean_valid[clean_valid["category"] != "unknown"].copy()
    bv = buggy_valid[buggy_valid["category"] != "unknown"].copy()

    all_cats = sorted(set(cv["category"].tolist() + bv["category"].tolist()))
    rng = np.random.default_rng(42)

    fig, ax = plt.subplots(figsize=(7, len(all_cats) * 0.7 + 1.2))

    for i, cat in enumerate(all_cats):
        y  = len(all_cats) - 1 - i
        cc = cv[cv["category"] == cat]["confidence"].values
        bc = bv[bv["category"] == cat]["confidence"].values

        if len(cc):
            ax.scatter(cc, [y + 0.15] * len(cc) + rng.uniform(-0.12, 0.12, len(cc)),
                       color=CLEAN_COLOR, s=55, alpha=0.85, zorder=3,
                       label="Clean" if i == 0 else "_")
            ax.plot([cc.mean()] * 2, [y + 0.02, y + 0.28],
                    color=CLEAN_COLOR, linewidth=2, zorder=4)

        if len(bc):
            ax.scatter(bc, [y - 0.15] * len(bc) + rng.uniform(-0.12, 0.12, len(bc)),
                       color=BUGGY_COLOR, s=55, alpha=0.85, zorder=3, marker="D",
                       label="Buggy" if i == 0 else "_")
            ax.plot([bc.mean()] * 2, [y - 0.28, y - 0.02],
                    color=BUGGY_COLOR, linewidth=2, zorder=4)

        ax.axhline(y + 0.5, color="#EEEEEE", linewidth=0.8, zorder=1)

    ax.set_yticks(range(len(all_cats)))
    ax.set_yticklabels([c.replace("_", " ") for c in reversed(all_cats)])
    ax.set_xlim(0, 1.05)
    ax.set_xlabel("Confidence")
    ax.set_title("Confidence by category — valid latents only\n"
                 "(circles = clean, diamonds = buggy; bar = mean)")
    ax.legend(frameon=False, loc="lower right")
    ax.axvline(0.5, color="#CCCCCC", linewidth=0.8, linestyle="--")
    ax.text(0.5, -0.6, "0.5", ha="center", va="top", fontsize=8, color="#999999")

    plt.tight_layout()
    for ext in ("pdf", "png"):
        fig.savefig(OUT_DIR / f"fig4_confidence_by_category.{ext}")
    plt.close(fig)
    print("Saved fig4")


# ══════════════════════════════════════════════════════════════════════════════
# Run all
# ══════════════════════════════════════════════════════════════════════════════

if __name__ == "__main__":
    fig1_category_comparison()
    fig2_latent_heatmap()
    fig3_valid_latents_table()
    fig4_confidence_by_category()
    print(f"\nAll figures saved to {OUT_DIR}")
    print("Use .pdf for the poster (vector), .png for quick preview.")
```
